This unit looks at active learning vs bayesian optimization.
- Active learning: Pick which point maximizes information gain
- Bayesian optimization: Pick which point we should evaluate next, if we want to optimize towards some goal

This is the motivation for combining epistemic term + pragmatic term.

**Acquisition functions** are those that select the next sample in bayesian optimization. The acquisition function is a function of the posterior $\alpha(x)$, and the argmax $argmax_x \alpha(x)$ is taken to select the next sample point.

There are many possible acquisition functions we can use in BO. Here are some.

**Probability of Improvement (PI)**: Chooses the next query point that has the highest *probability of improvement* over the current max $f(x^+)$.

Orange line is the current max (plus a small value). Violet region is probability density at each point. Grey shows probability density below the current max. The area of the violet region shows the "probability of improvement over current maximum". The next point to evaluate with PI is the dashed line, x = 6.



**Expected Improvement (EI)**: whereas PI looks only at *how likely* is an improvement, EI looks also at *how much* we improve. Choose the next query point that has the highest expected improvement over the current max $f(x^+)$.

**Thompson Sampling**: At every step, sample a function from the surrogate's posterior and optimize it. So like, sample a plausible distribution given the evidence and query wherever it peaks.

**Upper Confidence Bound (UCB)**: A linear combination of the mean and uncertainty of the surrogate model. Mean signifies exploitation (of the model's knowledge), and uncertainty signifies exploration (due to the model's lack of observations).

$$\alpha(x) = \mu(x) + \lambda \sigma(x)$$

The $\lambda$ value is the hyperparameter that can control the preference between exploration or exploitation.

**Gaussian Process Upper Confidence Bound (GP-UCB)**: This operationalizes *regret*. Imagine the current maximum is $a$ units, and we sample a point with $b < a$ units. Our regret is $a - b$. If we accumulate regret over $n$ iterations, we get **cumulative regret**.

$$\alpha_{GP-UCB}(x) = \mu_t(x) + \sqrt \beta_t \sigma_t(x)$$
